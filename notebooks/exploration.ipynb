# Exploratory Data Analysis for LLM Fine-Tuning Pipeline

This notebook demonstrates exploratory analysis of the fine-tuning dataset, inspecting record distributions, sequence lengths, ChatML formatting, and split hygiene.

```python
import json
import os
import pandas as pd
from pathlib import Path

train_path = Path('../data/processed/train.jsonl')
val_path = Path('../data/processed/validation.jsonl')
test_path = Path('../data/processed/test.jsonl')

def load_split(path):
    if not path.exists():
        return []
    with open(path, 'r', encoding='utf-8') as f:
        return [json.loads(line) for line in f if line.strip()]

train_data = load_split(train_path)
val_data = load_split(val_path)
test_data = load_split(test_path)

print(f"Train samples: {len(train_data)}")
print(f"Validation samples: {len(val_data)}")
print(f"Test samples: {len(test_data)}")
print(f"Total dataset size: {len(train_data) + len(val_data) + len(test_data)}")
```

### Sequence Length and Token Count Distributions

```python
if train_data:
    df = pd.DataFrame(train_data)
    df['instruction_char_len'] = df['instruction'].apply(len)
    df['response_char_len'] = df['response'].apply(len)
    df['total_char_len'] = df['text'].apply(len)
    df['approx_token_count'] = df['text'].apply(lambda x: len(x.split()) * 1.3)
    
    print(df[['instruction_char_len', 'response_char_len', 'approx_token_count']].describe())
```
